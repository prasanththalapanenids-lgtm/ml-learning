## import lib

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## dataset loading 

```python
dataset=pd.read_csv("Position_Salaries.csv")
X=dataset.iloc[:,1:-1].values
Y=dataset.iloc[:,-1].values
```

## now DataPreprocessing

```python
from sklearn.preprocessing import PolynomialFeatures
poly_feat=PolynomialFeatures(degree=4)
```

## Model Training

```python
from sklearn.linear_model import LinearRegression
x_poly_reg=poly_feat.fit_transform(X)
poly_reg=LinearRegression()
poly_reg.fit(x_poly_reg,Y)
```

## visualisation of Model

```python
plt.scatter(X,Y,color="black")
plt.plot(X,poly_reg.predict(poly_feat.fit_transform(X)),color="blue")
plt.xlabel("Level")
plt.title("Salary vs Level")
plt.ylabel("Salary")
plt.show()
```

## for smooth curve

```python
x_points=np.arange(min(X),max(X),0.1)
x_points=x_points.reshape(len(x_points),1)
plt.scatter(X,Y,color="black")
plt.plot(x_points,poly_reg.predict(poly_feat.fit_transform(x_points)),color="blue")
plt.title("Salary vs Level")
plt.xlabel("Level")
plt.ylabel("Salary")
plt.show()
```
