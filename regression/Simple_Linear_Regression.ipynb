## Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Dataset Loading

```python
dataset=pd.read_csv("Salary_Data.csv")
```

```python
dataset
```

```python
X=dataset.iloc[:,:-1].values
Y=dataset.iloc[:,-1].values
```

## Data Preprocessing

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=1)
```

## now training the model

```python
from sklearn.linear_model import LinearRegression
Reg=LinearRegression()
Reg.fit(x_train,y_train)
```

## Now predict the values

```python
## predict the remaining test data
y_prediction=Reg.predict(x_test)
```

```python
y_prediction
```

```python
## lets predict with a new value 
## the predict input always be in dataframe
Reg.predict([[11]])
```

## Visualising through the matplotlib

```python
plt.scatter(x_train,y_train,color="black")
plt.plot(x_train,Reg.predict(x_train),color="Green")
plt.title("Training set")
plt.xlabel("Years of experience")
plt.ylabel("Salary in USD")
plt.show()
```

```python
plt.scatter(x_test,y_test,color="black")
plt.plot(x_train,Reg.predict(x_train),color="Green")
plt.title("Model Prediction")
plt.xlabel("Years of experience")
plt.ylabel("Salary in USD")
plt.show()
```

### model testing metrics

```python
from sklearn.metrics import mean_absolute_error,root_mean_squared_error,r2_score
print(mean_absolute_error(y_test,y_prediction))
print(root_mean_squared_error(y_test,y_prediction))
print(r2_score(y_test,y_prediction))
```
